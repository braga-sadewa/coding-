## Part 3: Building the Budget Tracker

Now for the real thing. Ask AI to help you write a program that:
- Takes the user's monthly income
- Takes up to 3 expenses
- Calculates what's left over
- Warns the user if the budget goes negative

> 💡 **Pro tip:** The more specific your prompt, the less clean-up you'll need to do afterwards. "Write a budget tracker" gets you something generic; describing the exact inputs, outputs, and edge cases gets you something closer to what you actually need.


```python
# Write your detailed prompt here:
"""
My Budget Tracker Prompt:
Okay, now I want you to write another program to create a budget tracker. This program has several features, such as taking the user's monthly income, taking up to 3 expenses, and calculating what's left over. Additionally, it also needs to warn the user if the budget goes negative.
"""
```

### Paste and Test Your AI-Generated Budget Tracker

```python
# AI-GENERATED BUDGET TRACKER CODE
# Step 1: Get the monthly income
income = float(input("Enter your monthly income: $"))

# Step 2: Track expenses using a loop (up to 3 expenses)
total_expenses = 0.0

print("\nEnter your top 3 monthly expenses (type 0 if you have fewer than 3):")

for i in range(1, 4):
    expense = float(input(f"Enter expense #{i}: $"))
    total_expenses += expense  # Adds the expense to our running total

# Step 3: Calculate remaining balance
remaining_balance = income - total_expenses

# Step 4: Display the summary
print("\n" + "=" * 40)
print("           BUDGET SUMMARY           ")
print("=" * 40)
print(f"Total Income:     ${income:,.2f}")
print(f"Total Expenses:   ${total_expenses:,.2f}")
print(f"Remaining:        ${remaining_balance:,.2f}")
print("-" * 40)

# Step 5: Conditional warning if the balance is negative
if remaining_balance < 0:
    print("⚠️ WARNING: You are over budget! You spent more than you earned.")
elif remaining_balance == 0:
    print("💡 Notice: You broke even this month!")
else:
    print("🎉 Great job! You spent less than your income.")

print("=" * 40)
```

## Part 4: Enhancement Challenge

Pick ONE feature to add with AI's help:
- 🏷️ Expense categories (food, transport, entertainment)
- 📊 Percentage of income spent
- ✅ Input validation (no negative numbers)
- 📝 A detailed budget summary with advice


```python
# Document your enhancement choice and the prompt you used
enhancement_choice = "I chose to add: expense categories"
enhancement_prompt = """
Okay, so I tested the code and so far it worked very well. Now, I want to add a new feature in each of the expense, which is 'expense type', examples include rent, food, entertainment, shopping, etc. Write the same program which adds this new feature.
"""

print(f"🚀 Enhancement: {enhancement_choice}")
```

```python
# ENHANCED BUDGET TRACKER
# Step 1: Get the monthly income
income = float(input("Enter your monthly income: $"))

# Step 2: Track expenses and their types
expenses = []  # Empty list to store details for each expense
total_expenses = 0.0

print("\nEnter your top 3 monthly expenses (e.g., Rent, Food, Entertainment):")

for i in range(1, 4):
    expense_type = input(f"\nEnter category for expense #{i} (or '0' if none): ")

    # Skip if user types '0' or 'none'
    if expense_type.strip().lower() in ['0', 'none', '']:
        continue

    expense_amount = float(input(f"Enter amount for {expense_type}: $"))

    # Save as a dictionary and add to our list
    expenses.append({"type": expense_type, "amount": expense_amount})

    # Update running total
    total_expenses += expense_amount

# Step 3: Calculate remaining balance
remaining_balance = income - total_expenses

# Step 4: Display the summary
print("\n" + "=" * 45)
print("             BUDGET SUMMARY             ")
print("=" * 45)
print(f"Total Income:         ${income:,.2f}")
print("-" * 45)
print("EXPENSE BREAKDOWN:")

if expenses:
    for item in expenses:
        # Prints type left-aligned with formatting
        print(f"  • {item['type'].title():<15} ${item['amount']:,.2f}")
else:
    print("  • No expenses recorded.")

print("-" * 45)
print(f"Total Expenses:       ${total_expenses:,.2f}")
print(f"Remaining Balance:    ${remaining_balance:,.2f}")
print("-" * 45)

# Step 5: Conditional warnings
if remaining_balance < 0:
    print("⚠️ WARNING: You are over budget! You spent more than you earned.")
elif remaining_balance == 0:
    print("💡 Notice: You broke even this month!")
else:
    print("🎉 Great job! You spent less than your income.")

print("=" * 45)
```
