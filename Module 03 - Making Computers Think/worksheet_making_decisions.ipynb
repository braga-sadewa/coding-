Now that you have done the thinking, this is where AI earns its keep. Notice the difference:

````python
# Implement your academic standing system here
def classify_gpa(gpa: float) -> str:
    # Check bounds first (optional safety check)
    if gpa > 4.0 or gpa < 0.0:
        return "Invalid GPA"

    # Sequence from highest to lowest
    if gpa >= 4.0:
        return "A+"
    elif gpa >= 3.7:
        return "A"
    elif gpa >= 3.5:
        return "A-"
    elif gpa >= 3.3:
        return "B+"
    elif gpa >= 3.0:
        return "B"
    elif gpa >= 2.7:
        return "B-"
    else:
        return "Below B-"


# --- Testing the Function ---
test_scores = [4.0, 3.85, 3.6, 3.4, 3.1, 2.8, 2.2, 4.5]

for score in test_scores:
    print(f"GPA: {score:<4} -> Status: {classify_gpa(score)}")
````

### Activity 3: Debug the broken classifier

Here is some code that looks fine but misbehaves. Do not fix it by guessing. Trace the logic first.

Run the cell below. It should give "Good Standing" for a GPA of 3.2. Does it?

````python
student_gpa = 3.2

if student_gpa >= 2.0:
    status = "Probation"
elif student_gpa >= 3.0:
    status = "Good Standing"
elif student_gpa >= 3.5:
    status = "Dean's List"
else:
    status = "Failing"

print(f"Status: {status}")
````

**Trace it, then fix it.**

- When `student_gpa = 3.2`, is `3.2 >= 2.0` true? What happens the moment a condition is true?
- Why does the order of the conditions matter here?

> **Your analysis** (double-click to edit):

```
The problem is:
The >= symbol is arranged in the wrong order. Hence, Python already thinks that 3.2 is >= 2.0, ignoring the rest of the codes in the next lines.

It happens because:
Python executes line by line from top to bottom. Hence, the order of the code is strictly important to get the desired results and conditions.
```

Now write your corrected version below.

````python
# Your fixed version here
student_gpa = 3.2

# 1. Check the highest benchmark first
if student_gpa >= 3.5:
    status = "Dean's List"
# 2. Check the middle benchmark next
elif student_gpa >= 3.0:
    status = "Good Standing"
# 3. Check the minimum passing benchmark
elif student_gpa >= 2.0:
    status = "Probation"
# 4. Fallback for anything below 2.0
else:
    status = "Failing"

print(f"Status: {status}")
````

### Activity 5: Build an admissions decision system

Universities weigh several factors to make admission decisions. Design and build a fair, logical version.

**Design first:**

1. Choose 3 to 4 factors (GPA, test scores, activities, essay, special circumstances)
2. Decide how they combine (all required? at least 2 of 3?)
3. Set specific thresholds
4. Think about edge cases and tie-breakers

> **Your criteria** (double-click to edit):

```
Factor 1: GPA score
Factor 2: Activities participated
Factor 3: English test score
Decision logic:
All 3 factors play a role, with factor 1 and 3 play significant factor than the other 2 factors.
GPA score must be at least 2.5 or above for acceptance.
IELTS score must be at least 6.0 or above for acceptance.
Activities should be at least 2 or above, although not mandatory.
```

````python
# Your admissions decision system
# Tip: for "at least 2 of 3 conditions true", try counting how many are True.
def check_admission(gpa: float, ielts: float, activities: int) -> dict:
    """
    Evaluates university admission based on GPA, IELTS, and extracurricular activities.

    Mandatory Requirements:
      - GPA >= 2.5
      - IELTS >= 6.0

    Optional / Secondary Requirement:
      - Activities >= 2 (enhances acceptance standing)
    """
    # 1. Input Safety Checks
    if not (0.0 <= gpa <= 4.0):
        return {"accepted": False, "status": "Invalid GPA range"}
    if not (0.0 <= ielts <= 9.0):
        return {"accepted": False, "status": "Invalid IELTS score range"}
    if activities < 0:
        return {"accepted": False, "status": "Invalid activities count"}

    # 2. Hard Requirements Check (GPA and IELTS are strictly mandatory)
    meets_gpa = gpa >= 2.5
    meets_ielts = ielts >= 6.0
    has_strong_activities = activities >= 2

    # Both mandatory criteria must be True
    if meets_gpa and meets_ielts:
        if has_strong_activities:
            return {
                "accepted": True,
                "status": "Accepted (Strong Profile - High Activity Engagement)"
            }
        else:
            return {
                "accepted": True,
                "status": "Accepted (Regular - Meets Academic Requirements)"
            }
    else:
        # Collect failure reasons for clear feedback
        reasons = []
        if not meets_gpa:
            reasons.append(f"GPA {gpa} is below 2.5 requirement")
        if not meets_ielts:
            reasons.append(f"IELTS {ielts} is below 6.0 requirement")

        return {
            "accepted": False,
            "status": f"Rejected: {', '.join(reasons)}"
        }


# --- Test Cases ---
applicants = [
    {"name": "Alice", "gpa": 3.2, "ielts": 6.5, "activities": 3},  # Passes all
    {"name": "Bob",   "gpa": 2.8, "ielts": 6.0, "activities": 0},  # Passes mandatory, 0 activities
    {"name": "Charlie", "gpa": 2.2, "ielts": 7.0, "activities": 5},# Fails GPA
    {"name": "Diana", "gpa": 3.5, "ielts": 5.5, "activities": 2},  # Fails IELTS
]

for student in applicants:
    result = check_admission(student["gpa"], student["ielts"], student["activities"])
    print(f"Student: {student['name']:<7} | Result: {result['status']}")
````

## Key takeaways

**Computational thinking**
- Understand the problem before you code.
- Break decisions into clear rules.
- Test your logic with examples first.
- In an if/elif chain, order matters. Put the most specific conditions first.
- Use brackets to make complex conditions say what you mean. `and` binds tighter than `or`.

**Working with AI**
- AI is at its best when your requirements are clear.
- Use it to build your ideas, not to think for you.
- Give it your intent when you ask for help, especially when debugging.
