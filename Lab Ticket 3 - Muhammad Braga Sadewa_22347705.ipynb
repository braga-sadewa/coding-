# Note: This prompt is written in Gemini AI
```
Okay, I have this assignment to type in a single prompt for you to generate a code that runs perfectly when run in Python. So this is interesting and I am keen to see what you can come up with from me single-prompting the requirements I need. This can also be used by someone who just copy and paste this prompt to an AI tool like this. So basically, I want you to help a regular user to create a budget tracker for smart budget decisions. Now this budget tracker is supposed to track a user's budget for the month and informs the user if they have savings to spare, and if they are break-even, and also warns the user if they are overbudget. So I expect something like "what is your earnings this month?" and "what is your total expense for this month?". And I'm expecting something with some or of 'if' condition. It could be something like if the user's income is above expense, it provides output like 'congrats, your savings is [$$$]', so there is some sort of calculation involved as well in the final results displayed. And also if the income and expense is the same, something like 'you just hit break-even' for this month. And also, if the expense is bigger, then something like 'warning! you overspent [$$$]' this month. And I am also aware about some conditions that could be addressed. One of them is the total expense could be too vague. Maybe we can create major categories, such as 'rent', 'food', 'utilities', 'transport', 'entertainment' as major ones. But also putting 'others' for other expenses that are minor. And also, for the income, probably we can divide into something like 'salary' and 'other income' to add another different source in case the users have other sources of income. And also, noticed that this budget tracker is run for the month. So I want to make sure that it addressed what happens for the following month. Well, this is just a simple budget tracker that informs the user what will happen in a particular month, without having to connect it to the next months. So for instance, if the budget tracker provides a 'warning' due to expense > income, that should sufficiently informs the user about the course of action for the following month. And if the user receives a 'congrats' for having savings, then users are informed that they can utilize their savings from there. So yes, this budget tracker should have these inputs that users can provide, and provides outputs from these 3 conditions, which are savings remain, break even, and overspending. So yes, please provide this budget tracker for me to put into Python and I'm excited to see the results.
```

````python
# Paste the code your prompt produced here, then run it to show it works.
def budget_tracker():
    print("=" * 45)
    print("       MONTHLY SMART BUDGET TRACKER       ")
    print("=" * 45)

    # Helper function to get valid numeric input from the user
    def get_valid_amount(prompt):
        while True:
            try:
                amount = float(input(prompt))
                if amount < 0:
                    print("Amount cannot be negative. Please enter a valid number.")
                    continue
                return amount
            except ValueError:
                print("Invalid input. Please enter a valid numerical value.")

    # --- 1. INCOME INPUTS ---
    print("\n--- INCOME SOURCES ---")
    salary = get_valid_amount("Enter your salary for this month ($): ")
    other_income = get_valid_amount("Enter any other income for this month ($): ")
    total_income = salary + other_income

    # --- 2. EXPENSE INPUTS ---
    print("\n--- EXPENSE CATEGORIES ---")
    rent = get_valid_amount("Enter Rent / Housing ($): ")
    food = get_valid_amount("Enter Food / Groceries ($): ")
    utilities = get_valid_amount("Enter Utilities ($): ")
    transport = get_valid_amount("Enter Transportation ($): ")
    entertainment = get_valid_amount("Enter Entertainment ($): ")
    others = get_valid_amount("Enter Other expenses ($): ")

    total_expense = rent + food + utilities + transport + entertainment + others

    # --- 3. CALCULATIONS ---
    net_balance = total_income - total_expense

    # --- 4. SUMMARY DISPLAY ---
    print("\n" + "=" * 45)
    print("             MONTHLY SUMMARY             ")
    print("=" * 45)
    print(f"Total Income:   ${total_income:,.2f}")
    print(f"Total Expenses: ${total_expense:,.2f}")
    print("-" * 45)

    # --- 5. CONDITIONAL OUTPUTS & RECOMMENDATIONS ---
    # Condition 1: Surplus (Savings)
    if total_income > total_expense:
        print(f"STATUS: Congrats! You have savings of ${net_balance:,.2f} this month.")
        print("\nNext Month's Guidance:")
        print("- Consider putting your savings into an emergency fund or investments.")
        print("- You can carry over these savings to cushion future unexpected expenses.")

    # Condition 2: Break-Even
    elif total_income == total_expense:
        print("STATUS: You just hit break-even this month ($0.00 net balance).")
        print("\nNext Month's Guidance:")
        print("- You covered your costs, but you have no safety net built up.")
        print("- Aim to trim non-essential spending (e.g., entertainment) next month to build savings.")

    # Condition 3: Deficit (Overbudget)
    else:
        overspent_amount = abs(net_balance)
        print(f"STATUS: WARNING! You overspent by ${overspent_amount:,.2f} this month.")
        print("\nNext Month's Guidance:")
        print("- Review your largest expense categories to cut back immediately.")
        print("- Adjust your discretionary budget next month to pay off any debt incurred.")

    print("=" * 45)


# Run the application
if __name__ == "__main__":
    budget_tracker()
````
